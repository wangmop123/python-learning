Q6.
- WAP to read a number and check whether it is a palindrome number using loops and operators.

```python
num = int(input("Enter a number: "))
temp = num
rev = 0
while temp > 0:
    digit = temp % 10
    rev = rev * 10 + digit
    temp = temp // 10
if num == rev:
    print("Palindrome number")
else:
    print("Not a palindrome number")
```

Q7.
- WAP to read a number and check whether it is a prime number using loop and decision statements

```python
num = int(input("Enter a number: "))
prime = True
for i in range(2, num):
    if num % i == 0:
        prime = False
        break

if num > 1 and prime:
    print("Prime number")
else:
    print("Not a prime number")
```

1. Create a list and find the smallest and gratest element in list

```python
numbers = [12, 5, 8, 20, 3, 15]
smallest = numbers[0]
greatest = numbers[0]

for num in numbers:
    if num < smallest:
        smallest = num
    if num > greatest:
        greatest = num

print("Smallest number:", smallest)
print("Greatest number:", greatest)
    
    

```

2. Create a string with space and remove space from it.
- eg. text = "h e l l o w o r l d"

```python
text = "h e l l o w o r l d"
text_no_space = ""

for char in text:
    if char != " ":
        text_no_space += char

print(text_no_space)
```

##6. Create a dictionary Studentname and marks (5) name -> marks, Average marks

```python
students = {
    "pema":60,
    "dechen":80,
    "wangmo":78,
    "tashi":50,
    "tsepten":40
}
print("Student Marks:", students)
total = 0
for marks in students.values():
    total += marks

average = total / len(students)
print("Average Marks:", average)
    
```

```python
##class task
students = [
    ("Puja", [85, 90, 88]),
    ("Pratistha", [78, 69, 72]),
    ("Bijay", [92, 95, 90]),
    ("Norbu", [60, 65, 58])
]

student_avg = {}     
student_grade = {}    

highest_avg = -1
lowest_avg = 101
highest_name = ""
lowest_name = ""

class_total = 0

grade_count = {"A": 0, "B": 0, "C": 0}


for student in students:
    name = student[0]
    marks = student[1]

    total = 0
    for m in marks:
        total += m

    avg = total / len(marks)
    student_avg[name] = avg
    class_total += avg

    if avg >= 85:
        grade = "A"
    elif avg >= 70:
        grade = "B"
    else:
        grade = "C"

    student_grade[name] = grade
    grade_count[grade] += 1

   
    if avg > highest_avg:
        highest_avg = avg
        highest_name = name

    if avg < lowest_avg:
        lowest_avg = avg
        lowest_name = name
class_average = class_total / len(students)

print("Student Average:", student_avg)
print("Student Grades:", student_grade)

print("Highest Scorer:", highest_name, "Grade:", student_grade[highest_name])
print("Lowest Scorer:", lowest_name, "Grade:", student_grade[lowest_name])

print("Class Average:", class_average)
print("Grade Count:", grade_count)



    
```

### Inheritance :
- Inheritance is features of OOPS where the child class inherits the proporties of parent class
- Why ?
   - Code Reusability 
   - Code clean Architecture
   - Easy maintainence
   - Real world modeling



```python
# Parent class (Base class)

class Employee:
    def login(self):
        print("Login sucessfully")

## Single inheritance (Employee -> Developer)

class Developer(Employee):
    def code(self):
        print("Writes code")
# after creayting the class we need to create object
## Objects are always created for child class
dev = Developer()
dev.login() #inherited method from parent class
dev.code

## Why?
## One class extends another class
## Simple and most common
```

## multilevel inheritance
- Grandparent -> parent -> child

Employee -> TechnicalEmployee -> BackendDeveloper

```python
class TechnicalEmployee(Employee):
    def access_server(self):
        print("Acessing server")
        

class BackendDeveloper(TechnicalEmployee):
    def write_api(self):
        print("Writing API")
    
```

```python
backend = BackendDeveloper()
backend.login() ## grandparent class
backend.access_server() ##parent class
backend.write_api() ##own class


# technical = TechnicalEmployee()
# technical.login()
```

```python
### Hierarchical inheritance
- One parent --> multiple children
          Employee
        /          \
 Developer         Manager
```

```python

class Developer(Employee):
    def code(self):
        print("Writing code")

class Manager(Employee):
    def manage(self):
        print("managing team")
```

```python
ev = Developer()
mgr = Manager()

dev.login()
dev.code()

mgr.login()
mgr.manage()
```

```python
### Multiple Inheritance
     Employee              TeamLead
         \                   /
            senior developer
```

```python
class TeamLead:
    def lead(self):
        print("leading the team")


class SeniorDeveloper(Employee, TeamLead):
    def design_system(self):
        print("Designing System")
        
    
    
```

```python
senior = SeniorDeveloper()
senior.login()
senior.lead()
senior.design_system()
# ## Method resolution (MRO)
# print(SeniorDeveloper(mro))
```
