## Statistics using numpy

```python
arr = np.array([10,20,30,40,50,60])
print(np.mean(arr))
print(np.median(arr))
print(np.sum(arr))
print(np.std(arr))
print(np.var(arr))
print(np.min(arr))
print(np.max(arr))
print(np.sqrt(np.max(arr)))
```

```python
percentile_25 = np.percentile(arr,25)

percentile_75 = np.percentile(arr,75)

quartiles = np.percentile(arr,[25,50,70])
print(percentile_25,percentile_75,quartiles)
```

## Matrix Computation

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

print(f"Sum of two matrix: \n {A+B}")
print(f"Difference of two matrix: \n {A-B}")
print(f"Multiplication of two matrix: \n {A*B}")

#for dot product
print("Dot product:")
print(np.dot(A,B))

## transpose
print("Transpose:")
print(A.T)

## inverse
print("The inverse is",np.linalg.inv(A))

##determenant
print("The determenent of matrix A is",np.linalg.det(A))
```

## Sloving the linear equation

```python
''' 
  
   2x + 3y = 8
   5x + 7y = 19
'''
```

```python
cofficient = np.array([[2,3],[5,7]])
constants = np.array([8,19])

solution = np.linalg.solve(cofficient,constants)
print("The value are:",solution )
```

### Least square solution

- if there are more equation then unknowns then you cannot get exact solution so we calculate least square solution

```python
X = A(Transpose). A )inv . A(Transpose) B
```

```python
A = np.array([[1,1],[2,3],[4,5]])
B = np.array([2,5,8])


##Least squaire solution
x = np.linalg.lstsq(A,B, rcond=None)[0]
print(x)
```

## eigen value and eigen vector computation

```python
A = np.array([[1,1],[2,3]])
eigvec, eigval = np.linalg.eig(A)
print(eigvec,eigval)
```

```python
### Task1
Create employee data
Create a dictionary containing records of 10 employees. Each record must include the following fields:
* Name
* Age
* Position
* Salary
Convert to DataFrame
Convert the dictionary into a Pandas DataFrame. Display the DataFrame.
Salary analysis
Using the DataFrame, calculate and display:
* Average salary of all employees
* Minimum salary
* Maximum salary
Find highest paid employee
Identify the employee who has the maximum salary. Display:
* Employee name
* Position
* Salary
Bonus requirement, optional but recommended
Add a new column called Salary Category based on salary range:
* Low, salary less than 20000
* Medium, salary between 20000 and 35000
* High, salary above 35000
```

```python
import pandas as pd   

data = {
    'Name' : ['Ram','Sita','Gita','Pema','Shyam','Pemba','Norbu','Puja','Wangmo','Dechen'],
    'Age'  : [25, 24, 23, 21, 22, 30, 31, 32, 33, 34],
    'Position' : ['CEO','Manager','Clerk','HR','Staff','Staff','Staff','Staff','Staff','Staff'],
    'Salary' : [10000, 20000, 30000, 50000, 15000, 16000, 17000, 18000, 19000, 20000]
}

df = pd.DataFrame(data)

print("Employee DataFrame:")
print(df)


avg_salary = df['Salary'].mean()
min_salary = df['Salary'].min()
max_salary = df['Salary'].max()

print("\nAverage Salary:", avg_salary)
print("Minimum Salary:", min_salary)
print("Maximum Salary:", max_salary)


highest_paid = df[df["Salary"] == max_salary]

print("\nHighest Paid Employee:")
print("Name:", highest_paid.iloc[0]["Name"])
print("Position:", highest_paid.iloc[0]["Position"])
print("Salary:", highest_paid.iloc[0]["Salary"])


def salary_category(salary):
    if salary < 20000:
        return "Low"
    elif 20000 <= salary <= 35000:
        return "Medium"
    else:
        return "High"

df["Salary Category"] = df["Salary"].apply(salary_category)

print("\nDataFrame with Salary Category:\n")
print(df)
```

## Lets create a python script that generate datasets and save it to .csv file

```python
# import openpyxl
import numpy as np
import pandas as pd

#random seed
np.random.seed(42)

##Location
locations = ["Bhaktapur","Kathmandu","Lalitpur"]

#dictionary

data = {
    "id" :range(1,501),
    "area_sqft": np.random.randint(500,5000),
    "bedrooms": np.random.randint(1,6,500),
    "bathrooms": np.random.randint(1,5,500),
    "locations": np.random.choice(locations,500),
    "year_built": np.random.randint(1990,2026,500),
    "garage": np.random.randint(0,3,500),
    "price": np.random.randint(10000000,40000000,500)
}
df = pd.DataFrame(data)


## create some missing value in our datasets 

for col in ["area_sqft","bedrooms","garage","bathrooms","locations"]:
    df.loc[df.sample(frac= 0.01).index,col] = np.nan ## 5% values are missing

## define filename
house_csv = "houseDataset.csv"
# house_xlsx = "houseDatset.xlsx"

df.to_csv(house_csv, index= False)
# df.to_excel(house_xlsx, index= False)

print(f"{house_csv}  created sucessfully ")
    
```

```python
import matplotlib.pyplot as plt

categories = ['Basketball','Football','Cricket','Table Tennis']
values = [40,70,25,33]

plt.bar(categories, values, color=['red','blue','green','orange'], width=0.5,align='center', alpha=0.7)

# customization
plt.title("Students in Sports")
plt.xlabel("Categories")
plt.ylabel("values")
plt.grid(False)
for i, v in enumerate(values):
    plt.text(i, v, str(v), ha='center')

plt.figure(figsize=(7,5))
plt.tight_layout()
plt.show()
```
